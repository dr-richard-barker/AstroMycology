```python
import pandas as pd

# Path to your folder
folder_path = "/content/drive/MyDrive/Calibration_cm_3/"

# List to store filenames and volumes
volume_data = []

# Loop through files in the folder
for filename in os.listdir(folder_path):
    if filename.lower().endswith(".ply"):
        file_path = os.path.join(folder_path, filename)

        # Load mesh
        mesh = trimesh.load_mesh(file_path)

        # Calculate volume in mm³ and convert to cm³
        volume_mm3 = mesh.volume
        volume_cm3 = volume_mm3 / 1000  # 1 cm³ = 1000 mm³

        # Append to list
        volume_data.append({
            "File": filename,
            "Volume (mm³)": round(volume_mm3, 2),
            "Volume (cm³)": round(volume_cm3, 2)
        })

# Convert to DataFrame
df = pd.DataFrame(volume_data)

# Display table
print(df)
```
